2번 로또 당첨 여부 확인하기!

```python
import random

def generate_lotto_numbers():
    """랜덤으로 6개의 로또 번호를 생성하는 함수"""
    return sorted(random.sample(range(1, 46), 6))

def get_user_numbers():
    """사용자로부터 6개의 번호를 입력받는 함수"""
    while True:
        try:
            user_input = input("로또 번호 6개를 띄어쓰기로 입력하세요 (예: 1 12 23 34 41 45): ")
            user_numbers = list(map(int, user_input.split()))

            if len(user_numbers) != 6:
                print("⚠️ 6개의 숫자를 입력해야 합니다. 다시 입력해주세요.")
                continue

            if any(n < 1 or n > 45 for n in user_numbers):
                print("⚠️ 1~45 사이의 숫자만 입력 가능합니다. 다시 입력해주세요.")
                continue

            if len(set(user_numbers)) != 6:
                print("⚠️ 중복되지 않은 6개의 숫자를 입력해주세요. 다시 입력해주세요.")
                continue

            return sorted(user_numbers)

        except ValueError:
            print("⚠️ 숫자만 입력해주세요. 다시 입력해주세요.")

def check_lotto(user_numbers, winning_numbers):
    """사용자 번호와 당첨 번호를 비교하여 맞춘 개수를 반환"""
    matched_numbers = set(user_numbers) & set(winning_numbers)
    return len(matched_numbers), matched_numbers

# 로또 번호 생성
winning_numbers = generate_lotto_numbers()
print("🎉 이번 주 당첨 번호:", winning_numbers)

# 사용자 입력 받기
user_numbers = get_user_numbers()
print("🎟️ 내 번호:", user_numbers)

# 당첨 확인
match_count, matched_numbers = check_lotto(user_numbers, winning_numbers)
print(f"✅ 맞춘 개수: {match_count}개")

if match_count > 0:
    print(f"🎯 맞춘 번호: {sorted(matched_numbers)}")
else:
    print("😢 아쉽지만 당첨되지 않았어요.")
```

3번 번호 조합해서 막 만들어보기

```python
import random

def generate_lotto_numbers():
    """랜덤으로 6개의 로또 번호를 생성하는 함수"""
    return sorted(random.sample(range(1, 46), 6))

def generate_multiple_tickets(n):
    """사용자가 원하는 개수(n)만큼 로또 번호 조합을 생성하는 함수"""
    tickets = [generate_lotto_numbers() for _ in range(n)]
    return tickets

# 사용자 입력 받기
while True:
    try:
        num_tickets = int(input("몇 장의 로또를 구매하시겠습니까? "))
        if num_tickets < 1:
            print("⚠️ 1장 이상 구매해야 합니다. 다시 입력해주세요.")
            continue
        break
    except ValueError:
        print("⚠️ 숫자로 입력해주세요.")

# 로또 번호 생성
lotto_tickets = generate_multiple_tickets(num_tickets)

# 출력
print("\n🎟️ 생성된 로또 번호:")
for i, ticket in enumerate(lotto_tickets, start=1):
    print(f"  {i}번:", ticket)
```

수정본

```python
import random

def generate_lotto_numbers():
    """랜덤으로 6개의 로또 번호를 생성하는 함수"""
    return sorted(random.sample(range(1, 46), 6))

def simulate_lotto_draws(user_ticket, num_simulations):
    """주어진 로또 번호로 num_simulations번 로또를 구매하여 당첨 횟수를 분석"""
    win_count = {i: 0 for i in range(7)}  # 0~6개 맞춘 개수를 저장할 딕셔너리

    for _ in range(num_simulations):
        winning_numbers = generate_lotto_numbers()
        matched = len(set(user_ticket) & set(winning_numbers))
        win_count[matched] += 1  # 맞춘 개수별로 횟수 기록

    return win_count

# 사용자 로또 번호 입력 받기
def get_user_numbers():
    """사용자로부터 6개의 번호를 입력받는 함수"""
    while True:
        try:
            user_input = input("내 로또 번호 6개를 띄어쓰기로 입력하세요 (예: 1 12 23 34 41 45): ")
            user_numbers = list(map(int, user_input.split()))

            if len(user_numbers) != 6:
                print("⚠️ 6개의 숫자를 입력해야 합니다. 다시 입력해주세요.")
                continue

            if any(n < 1 or n > 45 for n in user_numbers):
                print("⚠️ 1~45 사이의 숫자만 입력 가능합니다. 다시 입력해주세요.")
                continue

            if len(set(user_numbers)) != 6:
                print("⚠️ 중복되지 않은 6개의 숫자를 입력해주세요. 다시 입력해주세요.")
                continue

            return sorted(user_numbers)

        except ValueError:
            print("⚠️ 숫자만 입력해주세요. 다시 입력해주세요.")

# 시뮬레이션 실행
user_ticket = get_user_numbers()
num_simulations = int(input("몇 번 로또를 시뮬레이션할까요? (예: 1000000): "))

win_stats = simulate_lotto_draws(user_ticket, num_simulations)

# 결과 출력
print(f"\n🎰 {num_simulations}번 로또를 시뮬레이션한 결과:")
for i in range(7):
    percentage = (win_stats[i] / num_simulations) * 100
    print(f"✅ {i}개 맞춤: {win_stats[i]}회 ({percentage:.6f}%)")
```
